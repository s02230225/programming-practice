Анализ задачи

Аналитически интеграл не берётся изученными методами, поэтому в качестве истинного значения будет значение, вычисленное встроенной функцией.

На отрезке [1, 2] интеграл собственный.

Второй отрезок вычисляется методами трапеции и Симпсона для n-2 отрезков, а 1 и 3 вычисляются отдельно. Для вычисления интеграла на 1 и последнем отрезках, используется формула Тейлора.


```python
# Импорт необходимых библиотек
import matplotlib.pyplot as plt
import numpy as np
import scipy as scipy
# Для установки в терминале pip install НАЗВАНИЕ
```

```python
# Определяем подынтегральную функцию
def f(x):
    return np.log(x + 2) / np.sqrt(3 * x - x ** 2)
```

```python
# Данo
a1 = 1
b1 = 2
a2 = 0
b2 = 3
left_lim = 0
right_lim = 3
n_val = [16, 32, 64]
```

Описание метода трапеций:

Метод трапеций — метод численного интегрирования функции одной переменной, заключающийся в замене на каждом отрезке подынтегральной функции на многочлен первой степени, то есть линейную функцию. Площадь под графиком функции аппроксимируется прямоугольными трапециями. Это простое применение формулы для площади трапеции — произведение полусуммы оснований, которыми в данном случае являются значения функции в крайних точках отрезка, на высоту (длину отрезка интегрирования).

В данном случае равномерной сетки формула может принять другой вид: x(i)=a+ih, где h=(b-a)/n - шаг сетки, тогда формула выглядит как полусумма f(a) и f(b) плюс f(x(i)) для i от 1 до n-1, умноженная на h.

```python
# Метод трапеций
def trapezoidal_rule(f, a, b, n):
    h = (b - a) / n
    if a == 0 and b == 3:
        a += h
        b -= h
        return 2*np.log(2)*np.arcsin(np.sqrt(1/n)) + (np.log(5))*(np.pi - 2*np.arcsin(np.sqrt(1-1/n))) + trapezoidal_rule(f, a, b, n - 2)
    result = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        x = a + i * h
        result += f(x)
    result *= h
    return result
```

```python
def output_trapezoidal(a, b, n):

    # Значение интеграла
    integral_approx = trapezoidal_rule(f, a, b, n)
    print(f"Приближенное значение интеграла (n={n}): {integral_approx:.6f}")

    # Создание графика
    plt.figure(figsize=(10, 6))
    plt.ylim(-1, 16)
    plt.axvline(x=a, color='orange', linestyle='--', label=f'x={a} и x={b}')
    plt.axvline(x=b, color='orange', linestyle='--')
    eps=0.0001
    x = np.linspace(a+eps, b-eps, 1000)
    y = f(x)

    # Разбиение графика трапециями
    if a==0 and b==3:
        a=a+(b-a)/n
        b=b-(b-a)/n
    x_trap = np.linspace(a, b, n + 1)
    y_trap = f(x_trap)
    plt.plot(x_trap, y_trap, 'r-', marker='o')

    # Отрисовка графика функции
    plt.plot(x, y, label='f(x)', color='blue', alpha=0.7)

    # Рисование сверху точками разбиения
    for j in range(n):
        plt.fill_between(x_trap[j:j + 2], y_trap[j:j + 2], alpha=0.3, color='green')

    # Отрисовка всего графика
    plt.title(f'Метод трапеций (n={n})')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.legend()
    plt.grid()
    plt.show()
```

```python
# Первый отрезок интегрирования
realin = scipy.integrate.quad(f, a1, b1)
print(f"Значение интеграла от {a1} до {b1}, вычисленное встроенной функцией {realin[0]} с точность {realin[1]}.")
```

Описание метода Симпсона:

Метод Симпсона — метод численного интегрирования функции одной переменной, заключающийся в замене на каждом отрезке подынтегральной функции на интерполяционный многочлен Лагранжа второй степени, то есть параболу. Площадь под графиком функции аппроксимируется параболами. Для метода необходимо выбрать обязательно чётное количество точек разбиения (n). 

В данном случае равномерной сетки формула может принять другой вид: x(i)=a+hi для i от 1 до n-1, где h=(b-a)/n, тогда формула принимает вид суммы f(a), f(b), 4f(x(i)) для нечётных i,  2f(x(i)) для чётных i, делённой на 3.

```python
# Метод Симпсона для нахождения площади
def simpsons_rule(f, a, b, n):
    h = (b - a) / n
    if a == 0 and b == 3:
        a += h
        b -= h
        return 2*np.log(2)*np.arcsin(np.sqrt(1/n)) + (np.log(5))*(np.pi - 2*np.arcsin(np.sqrt(1-1/n))) + simpsons_rule(f, a, b, n-2)
    result = f(a) + f(b)
    for i in range(1, n):
        x = a + i * h
        if i % 2 != 0:
            result += 4 * f(x)
        else:
            result += 2 * f(x)
    return result * h / 3
```

Для отрисовки парабол понадобится построить интерполяционный многочлен Лагранжа второй степени.

```python
# Строим интерполяционный многочлен Лагранжа второй степени
def lagrange_interpolation(x, y, x_val):
    n = len(x)
    result = 0.0
    for i in range(n):
        term = y[i]
        for j in range(n):
            if j != i:
                term *= (x_val - x[j]) / (x[i] - x[j])
        result += term
    return result
```

```python
def output_simpsons(a, b, n):

    # Значение интеграла
    integral_approx = simpsons_rule(f, a, b, n)
    print(f"Приближенное значение интеграла (n={n}): {integral_approx:.6f}")

    # Создание графика
    plt.figure(figsize=(10, 6))
    plt.ylim(-1, 16)
    plt.axvline(x=a, color='orange', linestyle='--', label=f'x={a} и x={b}')
    plt.axvline(x=b, color='orange', linestyle='--')
    eps=0.0001
    x = np.linspace(a+eps, b-eps, 1000)
    y = f(x)

    # Рисуем кривые Симпсона
    if a==0 and b==3:
        a=a+(b-a)/n
        b=b-(b-a)/n
    h = (b - a) / n
    for i in range(0, n, 2):
        x0 = a + i * h
        x1 = a + (i + 1) * h
        x2 = a + (i + 2) * h
        x_vals = np.array([x0, x1, x2])
        y_vals = f(x_vals)
        x_poly = np.linspace(x0, x2, 100)
        y_poly = np.array([lagrange_interpolation(x_vals, y_vals, val) for val in x_poly])
        plt.plot(x_poly, y_poly, color='red', alpha=0.7)
        plt.plot(x0, f(x0), marker='o', color='red', alpha=0.7)
        plt.plot(x1, f(x1), marker='o', color='red', alpha=0.35)
        plt.plot(x2, f(x2), marker='o', color='red', alpha=0.7)
        plt.fill_between(x_poly, y_poly, alpha=0.3, color='green')

    # Отрисовка графика функции
    plt.plot(x, y, label='f(x)', color='blue', alpha=0.7)

    # Отрисовка всего графика
    plt.title(f'Метод Симпсона (n={n})')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.legend()
    plt.grid()
    plt.show()
```
